```python
#pandas visualuzation
import pandas as pd
import matplotlib.pyplot as plt
#data
subjects = ["Math","physics","Chesmistry",'Biology',"English"]
marks = [85,90,78,88,76]
#plot
plt.plot(subjects,marks, marker='o',color='blue')
#customization
plt.title("Students marks in different Subjects")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.grid(True)
#show
plt.show()
```

```python
import matplotlib.pyplot as plt
days = ['mon','tue','wed','thu','fri','sat','sun']
tempurature = [30,32,31,29,33,35,34]
plt.plot(days,tempurature,markers='s',linestyle='--'color='red')
plt.title("Daily Tempurature")
plt.xlabel("Days")
plt.ylabel("Tempurature")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
month = ['jan','feb','mar','apr','may','jun']
sales = [15000,18000,12000,22000,24000,20000]
plt.bar(month,sales,color='green')
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
days = ['mon','tue','wed','thu','fri','sat','sun']
stock_price = [150,152,148,155,160,158,162]
plt.plot(days,stock_price,marker='o',color='blue')
plt.title("Daily Stock Price")
plt.xlabel("Days")
plt.ylabel("Stock Price")
plt.grid(True)
plt.show()
```
